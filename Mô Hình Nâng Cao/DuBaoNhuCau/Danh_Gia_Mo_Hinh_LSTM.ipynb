# BÁO CÁO ĐÁNH GIÁ MÔ HÌNH DỰ BÁO DOANH SỐ (LSTM)
**Đề tài:** AI-Powered Sales & Inventory Forecasting System
---

## 1. Kiến trúc Mô hình (Model Architecture)

Mô hình được xây dựng dựa trên mạng **Bidirectional LSTM (Long Short-Term Memory hai chiều)**. Đây là một dạng kiến trúc tiên tiến của mạng nơ-ron tái diễn (RNN).

*   **Lớp Bidirectional:** Cho phép mô hình nhìn dữ liệu theo cả hai chiều (tiến và lùi), giúp nắm bắt các quy luật lặp lại hiệu quả hơn.
*   **Cơ chế ghi nhớ dài hạn:** LSTM giải quyết vấn đề mất mát gradient (vanishing gradient), giúp học được các quy luật doanh số từ nhiều tháng trước.
*   **Đặc trưng đa biến (Multivariate):** Mô hình không chỉ dùng doanh số quá khứ mà còn kết hợp với các biến thời gian đã được mã hóa.

## 2. Các kỹ thuật Feature Engineering chuyên sâu

Để mô hình đạt hiệu năng cao, chúng tôi đã áp dụng các kỹ thuật xử lý dữ liệu đặc thù cho chuỗi thời gian:

1.  **Mã hóa Chu kỳ (Sine/Cosine Encoding):** 
    *   Chuyển đổi các cột Thứ (Day of Week) và Tháng (Month) thành tọa độ trên vòng tròn lượng giác.
    *   *Mục đích:* Giúp mô hình hiểu rằng Chủ Nhật và Thứ Hai nằm cạnh nhau, hay Tháng 12 và Tháng 1 là sự tiếp nối chu kỳ.
2.  **Đặc trưng Độ trễ (Lag Features):** 
    *   Sử dụng dữ liệu 30 ngày gần nhất (`look_back = 30`) để tạo bối cảnh dự báo.
3.  **Trung bình trượt (Moving Average):** 
    *   Thêm biến `rolling_mean_7` để mô hình nhận diện xu hướng tăng/giảm trong ngắn hạn.

## 3. So sánh LSTM với Random Forest (RF)

| Đặc điểm | Random Forest (RF) | LSTM (Của chúng ta) |
| :--- | :--- | :--- |
| **Bản chất** | Dựa trên các cây quyết định độc lập. | Dựa trên sự liên kết nơ-ron theo thời gian. |
| **Thứ tự thời gian** | Coi mỗi dòng dữ liệu là độc lập (không hiểu thứ tự). | Hiểu sâu sắc sự phụ thuộc giữa các ngày liên tiếp. |
| **Tính chu kỳ** | Cần rất nhiều đặc trưng thủ công để hiểu chu kỳ. | Tự học được các quy luật lặp lại (Seasonality). |
| **Hiệu năng** | Tốt trên tập dữ liệu tĩnh, nhỏ. | **Vượt trội trên dữ liệu chuỗi thời gian thực tế.** |

## 4. Ưu điểm và Hạn chế

### Ưu điểm (Pros):
*   **Độ chính xác cao:** Phù hợp với các cửa hàng bánh có biến động doanh số phức tạp.
*   **Tính thực tế:** Tích hợp sẵn bộ quy đổi sang nguyên liệu (bột mì, trứng...) giúp quản trị kho.
*   **Giả lập kịch bản:** Cho phép dự báo doanh số thay đổi thế nào khi giảm giá (Scenario Simulation).

### Hạn chế (Cons):
*   **Yêu cầu dữ liệu:** Cần lịch sử bán hàng ổn định để đạt kết quả tốt nhất.
*   **Tính 'Black-box':** Các trọng số bên trong mô hình khó giải thích một cách trực quan cho người không chuyên.

## 5. Các thước đo đánh giá thực tế (Business Metrics)

Thay vì chỉ nhìn vào Loss (MSE), mô hình sử dụng các thang đo dễ hiểu cho doanh nghiệp:
*   **MAE (Mean Absolute Error):** Sai số tuyệt đối trung bình (Ví dụ: lệch 5 sản phẩm/ngày).
*   **RMSE (Root Mean Squared Error):** Đo lường mức độ ổn định và các sai số lớn.
*   **R2 Score:** Thể hiện mức độ mà mô hình giải thích được sự biến động của thực tế (Càng gần 1 càng tốt).
